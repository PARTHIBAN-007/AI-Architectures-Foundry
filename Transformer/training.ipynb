```python
import torch
from transformers import AutoTokenizer
from model import Transformer, TransformerConfig
from tokenizer import FrenchTokenizer
config = TransformerConfig(
    embedding_dimension=512,
    num_attention_heads=8,
    mlp_ratio=4,
    encoder_depth=6,
    decoder_depth=6,
    src_vocab_size=30522,   
    tgt_vocab_size=32000    
)

device = "cuda" if torch.cuda.is_available() else "cpu"
src_tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")
tgt_tokenizer = FrenchTokenizer("trained_tokenizer/french_wp.json")

model = Transformer(config).to(device)
model.eval()

input_text = "I am learning about Large Language Models."
encoded_inputs = src_tokenizer(input_text, return_tensors="pt")
src_ids = encoded_inputs["input_ids"].to(device)  

with torch.inference_mode():
    output_token_ids = model.inference(
        src_ids=src_ids,
        tgt_start_id=2, 
        tgt_end_id=3,    
        max_len=50       
    )

decoded_text = tgt_tokenizer.decode(output_token_ids, skip_special_tokens=True)

print("Generated Token IDs:", output_token_ids)
print("Decoded Output Text:", decoded_text)
```
